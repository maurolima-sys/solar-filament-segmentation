```python
import os, subprocess

# Clone repo
subprocess.run(['git','clone','--depth','1',
    'https://github.com/maurolima-sys/solar-filament-segmentation.git',
    '/kaggle/working/repo'], check=True)

os.chdir('/kaggle/working/repo/report')

# Compile twice for references
for _ in range(2):
    r = subprocess.run(['pdflatex', '-interaction=nonstopmode', 'report.tex'],
                       capture_output=True, text=True)
    print(r.stdout[-3000:])
    if r.returncode != 0:
        print('STDERR:', r.stderr[-1000:])

import shutil
shutil.copy('report.pdf', '/kaggle/working/report.pdf')
print('PDF size:', os.path.getsize('/kaggle/working/report.pdf'), 'bytes')
```
